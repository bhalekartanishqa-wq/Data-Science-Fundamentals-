**Understanding Data Science Fundamentals**
Data Science is the discipline of extracting meaningful insights and actionable knowledge from data by combining statistics, programming, and domain expertise. It sits at the intersection of mathematics, computer science, and business understanding, allowing organizations to make informed decisions rather than relying on intuition alone.

**The Data Science Lifecycle**
The typical lifecycle follows these stages:
1.Problem Definition — Clearly understanding the business question or problem to solve. Without a well-defined problem, even the most sophisticated analysis won't produce useful results.
2.Data Collection — Gathering relevant data from databases, APIs, sensors, files, or web sources.
3.Data Cleaning & Preparation — Handling missing values, duplicates, inconsistent formats, and outliers. This stage often consumes the most time in a real project, since raw data is rarely analysis-ready.
4.Exploratory Data Analysis (EDA) — Investigating the data through summary statistics and visualizations to uncover patterns, trends, and relationships.
5.Modeling — Applying statistical or machine learning techniques to answer the problem, whether that's a prediction, classification, or clustering task.
6.Evaluation — Assessing how well the model or analysis performs, using appropriate metrics.
7.Deployment — Putting the solution into production so it can be used in real decision-making.
8.Monitoring & Maintenance — Tracking performance over time, since data patterns can shift ("data drift") and models may need retraining.

**Types of Data Analysis**
1.Descriptive Analysis — Summarizes what has happened in the past using metrics like averages, totals, and trends (e.g., "average monthly sales were $50,000").
2.Diagnostic Analysis — Investigates why something happened by looking for correlations and root causes (e.g., "sales dropped because of a supply chain delay").
3.Predictive Analysis — Uses historical data and statistical models to forecast future outcomes (e.g., predicting next quarter's revenue).
4.Prescriptive Analysis — Recommends specific actions to achieve a desired outcome, often using optimization techniques (e.g., suggesting the best pricing strategy).

**Common Industry Use Cases**
1.Healthcare — Predicting disease risk, optimizing treatment plans, and analyzing medical imaging.
2.Finance — Fraud detection, credit scoring, and algorithmic trading.
Retail & E-commerce — Recommendation systems, customer segmentation, and demand forecasting.
3.Sports — Performance analytics, injury prediction, and game strategy optimization (directly relevant to your other project).
4.Manufacturing — Predictive maintenance and quality control using sensor data.
5.Marketing — Customer churn prediction and targeted advertising.

**The Role of Data Science in Solving Real-World Problems**
Data Science transforms raw, often chaotic data into structured knowledge that drives decisions. Rather than guessing, businesses and organizations can validate hypotheses with evidence, identify hidden patterns humans might miss, and forecast future scenarios with quantified confidence. Its value lies not just in technical modeling, but in communicating findings clearly so non-technical stakeholders can act on them — making Data Science as much about storytelling and judgment as it is about code and math.

## Objective 2: Apply Python Programming Basics

This section demonstrates core Python concepts required for Data Science:
variables, data types, operators, control flow, functions, and collections.

### 2.1 Variables and Data Types

```
# Variables and basic data types
student_name = "Tanya"        # str
student_age = 19               # int
student_gpa = 7.35             # float
is_enrolled = True             # bool

print(type(student_name), type(student_age), type(student_gpa), type(is_enrolled))
```

### 2.2 Operators

```
# Arithmetic operators
a, b = 10, 3
print("Sum:", a + b)
print("Division:", a / b)
print("Modulus:", a % b)

# Comparison operators
print("Is equal:", a == b)
print("Is greater:", a > b)

# Logical operators
print("AND:", (a > 5) and (b > 5))
print("OR:", (a > 5) or (b > 5))
```

### 2.3 Conditional Statements

```
score = 78

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

### 2.4 Loops

```
# For loop: sum of numbers 1 to 5
total = 0
for i in range(1, 6):
    total += i
print("Sum using for loop:", total)

# While loop: countdown
count = 5
while count > 0:
    print("Countdown:", count)
    count -= 1
```

### 2.5 Functions

```
def calculate_average(numbers):
    """Returns the average of a list of numbers."""
    return sum(numbers) / len(numbers)

scores = [78, 85, 92, 66, 74]
print("Average score:", calculate_average(scores))
```

### 2.6 Lists, Tuples, Dictionaries, and Sets

```
# List - ordered, mutable
subjects = ["Math", "Science", "History"]
subjects.append("Art")
print("List:", subjects)

# Tuple - ordered, immutable
coordinates = (10.5, 20.3)
print("Tuple:", coordinates)

# Dictionary - key-value pairs
student_record = {"name": "Alice", "age": 21, "gpa": 3.75}
print("Dictionary:", student_record)

# Set - unordered, unique values only
unique_grades = {"A", "B", "A", "C", "B"}
print("Set (duplicates removed):", unique_grades)
```
